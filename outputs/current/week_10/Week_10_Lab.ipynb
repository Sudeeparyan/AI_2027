## 1. Set up the case

A synthetic invoice image has fields supplied by a prepared transcription.

```python
from decimal import Decimal
# Prepared transcriptions stand in for an image reader: this notebook does no OCR.
documents=[{'source':'IMG-01','qty':2,'unit':'20.00','tax':'5.00','listed':'54.00'},
           {'source':'IMG-02','qty':3,'unit':'12.00','tax':'0.00','listed':'36.00'}]
def check(d):
    calculated=d['qty']*Decimal(d['unit'])+Decimal(d['tax'])
    return {'source':d['source'],'calculated':calculated,'listed':Decimal(d['listed']),
            'matches':abs(calculated-Decimal(d['listed']))<=Decimal('0.01')}
for doc in documents: print(check(doc))
assert not check(documents[0])['matches']
```

Change IMG-01 listed total to 45.00. Name the distinction between a correctly transcribed value and a verified value.

## Inspect the synthetic invoice
This source illustration is generated locally from fictional values. The prepared transcription replaces OCR in this core exercise.

```python
def show_svg(svg):
    try:
        from IPython.display import display, SVG
        display(SVG(svg))
    except ImportError:
        print('Visual SVG prepared; open this notebook in Jupyter to display it.')
def grid(values, title):
    cells=''.join(f'<rect x="{20+c*32}" y="{45+r*32}" width="30" height="30" fill="rgb({int(max(0,min(1,values[r*4+c]))*255)}, {int(max(0,min(1,values[r*4+c]))*255)}, {int(max(0,min(1,values[r*4+c]))*255)})"/>' for r in range(4) for c in range(4))
    show_svg(f'<svg xmlns="http://www.w3.org/2000/svg" width="250" height="190"><rect width="250" height="190" fill="white"/><text x="12" y="25" font-size="15">{title}</text>{cells}</svg>')
```
