## 1. Set up the case

A document assistant must choose what earlier words matter when drafting a sentence.

```python
import math
def softmax(scores):
    top=max(scores)
    exps=[math.exp(x-top) if x != float('-inf') else 0 for x in scores]
    return [v/sum(exps) for v in exps]
scores=[1.0,2.0,0.0]
print('all visible:',[round(v,3) for v in softmax(scores)])
masked=[1.0,2.0,float('-inf')]
weights=softmax(masked)
print('future masked:',[round(v,3) for v in weights])
assert abs(sum(weights)-1)<1e-10 and weights[2]==0
```

The scores are supplied for a transparent calculation. They are not attention scores learned from a language model.
