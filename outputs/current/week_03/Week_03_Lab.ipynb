## 1. Set up the case

A college needs a synthetic poster image while respecting consent and image rights.

```python
import random
clean = [1.0 if r == 2 or c == 1 else 0.0 for r in range(4) for c in range(4)]
def noisy_image(strength, seed=7):
    rng = random.Random(seed)
    return [min(1,max(0,p + rng.uniform(-strength,strength))) for p in clean]
def mae(a,b): return sum(abs(x-y) for x,y in zip(a,b))/len(a)
for strength in (0.15,0.45):
    noisy=noisy_image(strength)
    # This toy has privileged access to clean: a real generator does not.
    toy_restored=[0.5*x+0.5*y for x,y in zip(noisy,clean)]
    print("noise",strength,"MAE noisy",round(mae(noisy,clean),3),"toy restored",round(mae(toy_restored,clean),3))
```

## See the denoising stages
The toy blends with a known clean target. This visualizes progressive restoration and is not a trained diffusion model.

```python
def show_svg(svg):
    try:
        from IPython.display import display, SVG
        display(SVG(svg))
    except ImportError:
        print('Visual SVG prepared; open this notebook in Jupyter to display it.')
def grid(values, title):
    cells=''.join(f'<rect x="{20+c*32}" y="{45+r*32}" width="30" height="30" fill="rgb({int(max(0,min(1,values[r*4+c]))*255)}, {int(max(0,min(1,values[r*4+c]))*255)}, {int(max(0,min(1,values[r*4+c]))*255)})"/>' for r in range(4) for c in range(4))
    show_svg(f'<svg xmlns="http://www.w3.org/2000/svg" width="250" height="190"><rect width="250" height="190" fill="white"/><text x="12" y="25" font-size="15">{title}</text>{cells}</svg>')
```

```python
noisy=noisy_image(.45)
for alpha in [0,.5,1]:grid([(1-alpha)*a+alpha*b for a,b in zip(noisy,clean)],f'Toy restoration {alpha:.1f}')
prepared_outputs=[{'id':'A','brief_match':True,'text_legible':False,'consent_ok':True},
                  {'id':'B','brief_match':True,'text_legible':True,'consent_ok':False}]
for result in prepared_outputs:
    print(result,'ready=',all(result[k] for k in ['brief_match','text_legible','consent_ok']))
print('Each criterion has a different reason. Explain which revision A and B need.')
```
