## 1. Set up the case

A student asks whether a fictional college reimburses a late bus ticket.

```python
questions = ["late bus", "renew library loan", "deadline extension", "appeal window", "travel costs", "book renewal", "two extensions", "appeal after twelve days", "staff payroll", "cafeteria menu"]
reference = ["no", "yes", "one", "ten days", "no", "yes", "no", "no", "staff only", "unsupported"]
answers = ["no", "yes", "one", "ten days", "no", "yes", "no", "no", "staff only", "unsupported"]
rows = [(q, expected, actual, expected == actual) for q, expected, actual in zip(questions, reference, answers)]
for row in rows: print(row)
print("Accuracy:", sum(row[-1] for row in rows), "/", len(rows))
assert len(rows) == 10
```
