## 1. Set up the case

An office extracts amounts and invoice identifiers from synthetic invoices.

```python
def validate(record):
    required={'invoice_id':str,'quantity':int,'unit_price':(int,float),'listed_total':(int,float,type(None))}
    errors=[f'{key}: missing or wrong type' for key,typ in required.items() if key not in record or not isinstance(record[key],typ)]
    if not errors and record['listed_total'] is not None:
        expected=record['quantity']*record['unit_price']
        if abs(expected-record['listed_total'])>0.01: errors.append('listed total differs from calculation')
    return errors
records=[{'invoice_id':f'INV-{i:02d}','quantity':i%3+1,'unit_price':12,'listed_total':(i%3+1)*12} for i in range(10)]
records[3]['listed_total']=None
for r in records: print(r['invoice_id'],validate(r), 'listed_total=',r['listed_total'])
assert validate(records[3])==[] and records[3]['listed_total'] is None
```

Missing total is explicitly None. Add a separate review flag for it and never label a computed value as a field extracted from the invoice.

## Before and after prompt, then evidence checks
Use the same ten cases for both prompt versions. Prepared records demonstrate checks; prompt wording alone does not guarantee correct extraction.

```python
before='Read this invoice.'
after='Extract invoice_id, quantity, unit_price and listed_total from the supplied invoice only. Return null for absent fields. Do not infer the listed total.'
print('BEFORE:',before,'\nAFTER:',after)
for r in records:
    review=r.get('listed_total') is None
    print(r['invoice_id'],'schema_errors=',validate(r),'missing_field_review=',review)
assert records[3]['listed_total'] is None
```
