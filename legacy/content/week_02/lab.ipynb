## 1. Set up the case

A designer wants new simple icons similar to examples, without copying an exact icon.

```python
# Deterministic toy decoder; no training, no encoder distribution, no discriminator.
def toy_decode(z):
    assert 0 <= z <= 1
    return [round(z, 2), round(1-z, 2), round(z*z, 2)]
target = [0.3, 0.7, 0.09]
for z in (0.3, 0.7):
    result = toy_decode(z)
    mae = sum(abs(a-b) for a,b in zip(target,result))/len(target)
    print("z",z,"reconstruction",result,"MAE",round(mae,3))
```

```python
def show_svg(svg):
    try:
        from IPython.display import display, SVG
        display(SVG(svg))
    except ImportError:
        print('Visual SVG prepared; open this notebook in Jupyter to display it.')
def grid(values, title):
    cells=''.join(f'<rect x="{20+c*32}" y="{45+r*32}" width="30" height="30" fill="rgb({int(max(0,min(1,values[r*4+c]))*255)}, {int(max(0,min(1,values[r*4+c]))*255)}, {int(max(0,min(1,values[r*4+c]))*255)})"/>' for r in range(4) for c in range(4))
    show_svg(f'<svg xmlns="http://www.w3.org/2000/svg" width="250" height="190"><rect width="250" height="190" fill="white"/><text x="12" y="25" font-size="15">{title}</text>{cells}</svg>')
```

```python
train=[.30,.18,.12];validation=[.32,.20,.26]
lines=[]
for name,vals,color in [('training',train,'blue'),('validation',validation,'orange')]:
    pts=' '.join(f'{60+i*130},{230-v*500}' for i,v in enumerate(vals))
    lines.append(f'<polyline fill="none" stroke="{color}" stroke-width="3" points="{pts}"/>')
show_svg('<svg xmlns="http://www.w3.org/2000/svg" width="450" height="280"><rect width="450" height="280" fill="white"/><text x="20" y="20">Supplied loss: blue=train, orange=validation</text><text x="20" y="45">Error</text><path d="M60 50V230H340" fill="none" stroke="black"/>'+''.join(lines)+'<text x="70" y="255">Epoch 1        Epoch 2        Epoch 3</text></svg>')
def icon(z):
    return [min(1,max(0,(0.9 if r==2 or c==1 else 0.1)+(z-.5)*(.3 if c>1 else -.3))) for r in range(4) for c in range(4)]
for z in [.3,.7]:grid(icon(z),f'Toy code z={z}')
print('Training errors:',train,'Validation errors:',validation)
print('Task: label encoder/sample/decoder; then label generator/discriminator/feedback.')
```
